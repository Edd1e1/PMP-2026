```python
import numpy as np

def experiment():
    urna = ['R'] * 3 + ['A'] * 4 + ['N'] * 2
    zar = np.random.randint(1, 7)

    if zar in [2, 3, 5]:
        urna.append('N')
    elif zar == 6:
        urna.append('R')
    else:
        urna.append('A')

    return np.random.choice(urna)

N = 100_000
rosii = sum(1 for _ in range(N) if experiment() == 'R')
p_estimat = rosii / N

p_teoretic = (3/6 * 3/10) + (1/6 * 4/10) + (2/6 * 3/10)

print(f"Probabilitate estimata: {p_estimat:.5f}")
print(f"Probabilitate teoretica: {p_teoretic:.5f}")
print(f"Diferenta: {abs(p_estimat - p_teoretic):.5f}")
```
