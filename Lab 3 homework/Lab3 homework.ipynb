```python
import numpy as np

N = 100_000
X = np.random.randint(1, 7, N)
Y = np.random.randint(1, 7, N)
S = X + Y

A = S >= 10
B = X % 2 == 0
C = (X % 2 == 0) | (Y % 2 == 0)
D = (X % 2 == 0) & (Y % 2 == 0)
E = S == 8

def cond(ev, given):
    return (ev & given).sum() / given.sum()

print(cond(A, B))
print(cond(A, C))
print(cond(D, E))
```

```python
import pandas as pd

df = pd.DataFrame({
    "tratament": ["T1", "T1", "T2", "T2"],
    "severitate": ["usoara", "severa", "usoara", "severa"],
    "vindecati": [190, 30, 48, 130],
    "total": [200, 50, 50, 200],
})

df["rata"] = df["vindecati"] / df["total"]
print(df.pivot(index="severitate", columns="tratament", values="rata"))

glob = df.groupby("tratament")[["vindecati", "total"]].sum()
glob["rata"] = glob["vindecati"] / glob["total"]
print(glob)
```
